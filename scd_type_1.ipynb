```python
%sql
create table if not exists ecommerce_dwh.data_marts.dim_customer_scd
as select * from ecommerce_dwh.data_marts.dim_customer
where 1=0
```

```python
%sql
create or replace view ecommerce_dwh.data_marts.dim_customer_view as
select * from ecommerce_dwh.data_marts.dim_customer_scd
```
